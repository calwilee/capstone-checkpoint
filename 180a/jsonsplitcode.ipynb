```python
import json
import random
import math
```

```python
# Path to your original JSON file
input_path = "ambiginstfull.json"

# Output file path for the sampled subset
output_path = "ambiginst.json"

# Percentage of items to sample (e.g., 40 means 40%)
sample_percentage = 20

# --- LOAD JSON FILE ---
with open(input_path, "r", encoding="utf-8") as f:
    data = json.load(f)

# --- SAMPLE LOGIC ---
n_total = len(data)
n_sample = math.ceil(n_total * sample_percentage / 100)

sampled_data = random.sample(data, n_sample)

# --- SAVE TO NEW JSON FILE ---
with open(output_path, "w", encoding="utf-8") as f:
    json.dump(sampled_data, f, indent=4, ensure_ascii=False)

# --- DISPLAY INFO ---
print(f"✅ Sampled {n_sample} out of {n_total} entries ({sample_percentage}%).")
print(f"💾 Saved to {output_path}")
```
