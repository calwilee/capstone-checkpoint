```python
import requests, os, json
OLLAMA_HOST = os.environ.get("OLLAMA_HOST", "http://localhost:11434")

# Ping the server
print("GET /api/tags ->", requests.get(OLLAMA_HOST.rstrip("/") + "/api/tags", timeout=5).status_code)

# Minimal chat smoke test
resp = requests.post(
    OLLAMA_HOST.rstrip("/") + "/api/chat",
    json={"model":"llama3.1","messages":[{"role":"user","content":"Reply with JSON: {\"ok\":true}"}],"stream":False},
    timeout=30
)
print("POST /api/chat ->", resp.status_code, resp.json().get("message",{}).get("content","")[:120])
```
