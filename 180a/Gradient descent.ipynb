```python
import numpy as np
import seaborn as sns 
import matplotlib as plt
import pandas as pd
```

```python
data = pd.read_csv('HW2_ellipse.csv', delimiter = ',',header=None)
sns.scatterplot(x = data[0], y = data[1])
```

```python
A = data.apply(lambda x: x**2)
A = A.values
b = np.ones(1000)

def f(a):
    s = A @ a - b
    return np.dot(s,s)

def grad_f(a):
    return 2 * A.T @ (A @ a - b)

n = 1 / np.linalg.norm(A.T @ A ,2) / 2

a = [1,1]                    
tol = 1e-10

for i in range(100000):
    g = grad_f(a)
    if np.linalg.norm(g) < tol:
        break
    a = a - n * g
a
```

```python
def pelip(x):
    return ((1 - a[0] * x**2) / a[1])**.5
def nelip(x):
    return -((1 - a[0] * x**2) / a[1])**.5
sns.scatterplot(x = data[0], y = data[1])

x_values = np.linspace(-2,2, 1000) 
sns.lineplot(x = x_values,y =  pelip(x_values), color = 'red')
sns.lineplot(x = x_values,y =  nelip(x_values), color = 'red')
```
