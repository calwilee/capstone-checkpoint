```python
def backtrack(f, g, x, d, mu0=1.0, beta=0.8, gamma=0.2):
    
    fx = f(x)
    g = g(x)
    mu = mu0
    # Armijo condition
    while f(x + mu * d) > fx + gamma * mu * np.dot(g, d):
        mu *= beta
    return mu

```

```python
def f(x):
    x1, x2 = x
    return (2*x1 - 1)**4 + (x1 + x2 - 1)**2
    
def g_f(x):
    x1, x2 = x
    return np.array([8*(2*x1 - 1)**3 + 2*(x1 + x2 - 1), 2*(x1 + x2 - 1)])
    
def backtrack_desc(mu, gamma, beta):
    x_0 = np.array([0,0])
    grad_x_0 = g_f(x_0)
    x_1 = x_0 - mu * g_f(x_0)

    # checks Armijo’s
    while f(x_1) > f(x_0) - mu * gamma * np.dot(grad_x_0, grad_x_0):
        mu *= beta
        x_1 = x_0 - mu * grad_x_0
        
    return mu
    
backtrack_desc(1, .2, .8)
```

```python
import numpy as np

# --- your objective and gradient ---
def f(x):
    x1, x2 = x
    return (2*x1 - 1)**4 + (x1 + x2 - 1)**2

def grad(x):
    x1, x2 = x
    return np.array([
        8*(2*x1 - 1)**3 + 2*(x1 + x2 - 1),
        2*(x1 + x2 - 1)
    ], dtype=float)

# --- backtracking (Armijo) ---
def backtracking_line_search(f, grad, x, d, mu0=1.0, beta=0.5, gamma=0.2):
    fx = f(x)
    g = grad(x)
    mu = mu0
    while f(x + mu*d) > fx + gamma * mu * np.dot(g, d):
        mu *= beta
    return mu

# ----- (b)(ii) at x^(0) = (0,0) -----
x0 = np.array([0.0, 0.0])
g0 = grad(x0)
d0 = -g0              # steepest descent direction
gamma = 0.2
beta = 0.8
mu0 = 1.0

mu = backtracking_line_search(f, grad, x0, d0, mu0=mu0, beta=beta, gamma=gamma)
x1 = x0 + mu * d0

# Verify Armijo inequality:
lhs = f(x1)
rhs = f(x0) + gamma * mu * np.dot(g0, d0)  # note dot(g0, d0) = -||g0||^2

print(f"Accepted step size mu = {mu}")
print(f"x^(1) = {x1}")
print(f"Armijo check: f(x^(1)) <= f(x^(0)) - gamma*mu*||grad||^2 ?")
print(f"LHS = {lhs}")
print(f"RHS = {rhs}")
print("Holds?", lhs <= rhs)
```
